```python
import pandas as pd
```

```python
data = pd.read_csv("./Dataset/gold_price_data.csv")
df = pd.DataFrame(data)
df.info()
```

```python
df.describe().T
```

```python
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'],format = "%m/%d/%Y")
df['Day'] = df['Date'].dt.day
df.head()
```

```python
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# plt.figure(figsize =(8,6))
# plt.title("Trend of GLD over Time(Day)")
# plt.plot(df["Month"],df["GLD"],)
# plt.xlabel("GLD variation over time")
# plt.ylabel("Day")
# plt.show()

df['GLD'].plot()
```

```python
correlation =  df[["SPX","GLD","USO","SLV","EUR/USD"]].corr()
sns.heatmap(correlation, annot = True , fmt = ".2f")
plt.plot()
```

```python
df["GLD_lag_1"] = df["GLD"].shift(1)
df["GLD_lag_2"] = df["GLD"].shift(2)
df["GLD_lag_3"] = df["GLD"].shift(3)

df.head(10)
```

```python
df.dropna(inplace = True)
df.head()
```

```python
df.drop(df[['Date','Day']], inplace = True, axis = 1)
```

```python
df.head(20)
```

```python
df.info()
```

```python
x = df.drop(['GLD'],axis=1)
y = df['GLD']
```

```python
sns.heatmap(x.corr(), annot = True , fmt = ".2f")
plt.plot()
```

```python
from sklearn.model_selection import train_test_split 
from sklearn.ensemble import RandomForestRegressor

# x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2 , random_state = 42)

split = int(len(x) * 0.8)

x_train = x[:split]
x_test = x[split:]

y_train = y[:split]
y_test = y[split:]

x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
rfr = RandomForestRegressor()
rfr.fit(x_train,y_train)
y_preds = rfr.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

print("R2:", r2_score(y_test, y_preds))
print("MAE:", mean_absolute_error(y_test, y_preds))
```
